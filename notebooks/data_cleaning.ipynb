```python
# Remove listings with zero price
df_clean = df_clean[df_clean["price"] > 0].copy()

# Replace the clearly invalid maximum_nights value
df_clean.loc[
    df_clean["maximum_nights"] == 2147483647,
    "maximum_nights"
] = np.nan

print("Shape after invalid-value cleaning:", df_clean.shape)
```

```python
print("Zero-price listings:", (df_clean["price"] == 0).sum())
print("Invalid maximum_nights:",
      (df_clean["maximum_nights"] == 2147483647).sum())
```

```python
print(df_clean["price"].describe())
```

```python
# Calculate IQR for price

Q1 = df_clean["price"].quantile(0.25)
Q3 = df_clean["price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

# Count potential outliers
price_outliers = df_clean[
    (df_clean["price"] < lower_bound) |
    (df_clean["price"] > upper_bound)
]

print("Number of potential price outliers:", len(price_outliers))
```

```python
price_outliers = df_clean[
    df_clean["price"] > upper_bound
]

price_outliers[[
    "id",
    "name",
    "neighbourhood_cleansed",
    "room_type",
    "accommodates",
    "bedrooms",
    "bathrooms",
    "price"
]].sort_values("price", ascending=False).head(20)
```
